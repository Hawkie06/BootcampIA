```python
plt.figure(figsize=(10, 5))

sb.countplot(
    data=data,
    x="Faixa salarial mensal:"
)

plt.title("Distribuição das faixas salariais")
plt.xlabel("Faixa salarial")
plt.ylabel("Quantidade de pessoas")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sb.countplot(
    data=data,
    x="Escolaridade:",
    hue="Faixa salarial mensal:"
)

plt.title("Escolaridade × Faixa salarial mensal")
plt.xlabel("Escolaridade")
plt.ylabel("Quantidade de pessoas")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Faixa salarial")

plt.tight_layout()
plt.show()
```
