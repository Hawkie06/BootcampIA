## Criando um DataFrame

Um `DataFrame` organiza dados em linhas e colunas.

````python
# As chaves do dicionário são os nomes das colunas, as listas são valores
dados = {
    "nome": ["Ana", "Bruno", "Carlos", "Daniela", "Eduardo", "Gustavo"],
    "idade": [22, 25, 21, 28, 24, 23],
    "curso": [
        "Python",
        "Python",
        "Machine Learning",
        "Python",
        "Machine Learning",
        "Machine Learning"
    ],
    "nota": [8.5, 7.0, 9.2, 6.5, 8.0, 9.0],
    "aprovado": [True, True, True, False, True, True]
}

data_frame = pd.DataFrame(dados)

data_frame
````

````python
# Carregando dados lendo arquivo
#data_frame = pd.read_csv('caminho_do_arquivo')

print(data_frame)
````

### Estrutura do DataFrame

- **Colunas:** características dos dados, como nome, idade e nota.
- **Linhas:** cada registro, neste caso, cada aluno.
- **Índice:** identifica cada linha. Por padrão, começa em `0`.
- **DataFrame:** estrutura que reúne todas essas informações.

## Filtrando registros

### Nota maior ou igual a 8

````python
# Utiliza operação de comparação para filtrar os valores
data_frame[data_frame["nota"] >= 8]
````

### Filtrando por texto

````python
data_frame[data_frame["curso"] == "Python"]
````

### Combinando condições

````python
# Utiliza operadores lógicos para combinar condições
data_frame[
    (data_frame["curso"] == "Python") &
    (data_frame["nota"] >= 8)
]
````

Operadores importantes:

| Operador | Significado |
|---|---|
| `==` | Igual |
| `!=` | Diferente |
| `>` | Maior |
| `<` | Menor |
| `>=` | Maior ou igual |
| `<=` | Menor ou igual |
| `&` | E |
| `\|` | Ou |
| `~` | Negação |

Ao combinar condições no Pandas, coloque cada condição entre parênteses.

### Selecionando com `loc`

````python
# Primeiro índice é o filtro e as segundo as colulas desejadas
data_frame.loc[data_frame["nota"] >= 8, ["nome", "nota"]]
````

## Criando e alterando colunas

### Criando uma coluna booleana

````python
# Criação de coluna é semelhante à criação de nova chave no dicionário
data_frame["situacao"] = data_frame["nota"] >= 7

data_frame
````

### Criando uma coluna textual com `numpy.where`

````python
import numpy as np

# Primeiro parâmetro é a condição, segundo a atribuição caso verdadeira e terceira caso falsa
data_frame["resultado"] = np.where(
    data_frame["nota"] >= 7,
    "Aprovado",
    "Reprovado"
)

data_frame
````

### Alterando uma coluna

````python
# Criando uma cópia para preservar o DataFrame original.
df_copia = data_frame.copy()

# Atribui a transformação para todos os valores da coluna selecionada
df_copia["nota"] = df_copia["nota"] + 0.5

df_copia
````

### Removendo colunas

````python
data_frame = data_frame.drop(columns=["situacao", "resultado"])

data_frame
````

## Calculando estatísticas

````python
print("Média:", data_frame["nota"].mean())
print("Maior nota:", data_frame["nota"].max())
print("Menor nota:", data_frame["nota"].min())
print("Soma:", data_frame["nota"].sum())
print("Mediana:", data_frame["nota"].median())
print("Quantidade de notas:", data_frame["nota"].count())
````

## Agrupando dados

O `groupby()` permite analisar registros separados por categorias.

### Média das notas por curso

````python
data_frame.groupby("curso")["nota"].mean()
````

### Quantidade de alunos por curso

````python
data_frame.groupby("curso")["nome"].count()
````

````python
data_frame["curso"].value_counts()
````

### Várias estatísticas por grupo

````python
data_frame.groupby("curso")["nota"].agg(
    ["count", "mean", "min", "max"]
)
````

### Resultado como DataFrame

````python
media_por_curso = data_frame.groupby("curso", as_index=False)["nota"].mean()

media_por_curso
````

## Ordenando dados

### Ordem crescente

````python
# Informa nome da coluna
data_frame.sort_values("nota")
````

### Ordem decrescente

````python
# Define a forma de ordenar
df_ordenado = data_frame.sort_values("nota", ascending=False)

df_ordenado
````

### Ordenando por várias colunas

````python
df_ordenado = data_frame.sort_values(
    by=["curso", "nota"],
    ascending=[True, False]
)

df_ordenado
````

### Ordenando e redefinindo o índice

````python
df_ordenado = data_frame.sort_values("nota", ascending=False).reset_index(drop=True)

df_ordenado
````

## Trabalhando com valores ausentes

Em dados reais, algumas informações podem estar ausentes. O Pandas utiliza, entre outras representações, `NaN` para esses valores.

````python
df_nulos = data_frame.copy()
df_nulos.loc[2, "nota"] = None

df_nulos
````

### Identificando valores ausentes

````python
df_nulos.isna()
````

````python
df_nulos.isna().sum()
````

### Removendo registros com valores ausentes

````python
df_nulos.dropna()
````

````python
df_nulos.dropna(subset=["nota"])
````

### Preenchendo valores ausentes

````python
media = df_nulos["nota"].mean()

df_preenchido = df_nulos.copy()
df_preenchido["nota"] = df_preenchido["nota"].fillna(media)

df_preenchido
````

A escolha entre remover e preencher dados ausentes depende do significado dos dados e do objetivo da análise.
